```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

**Упражнение 1**



$y_1 = y_2$

$2x=x^2$

$x=0$ или $x=2$

```python
x = np.linspace(0, 2, 100)
y1 = 2 * x
y2 = x**2
plt.plot(x, y1, label='y = 2x')
plt.plot(x, y2, label='y = x^2')
plt.fill_between(x, y1, y2, where=(y1 > y2), color='lightblue', alpha=0.5)
plt.title('Region between y = 2x and y = x^2')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()

x, y = sp.symbols('x y')
f = x + y**2
c, d = (x**2, 2*x)
a, b = (0, 2)
Iy = sp.integrate(f, (y, c, d))
Ix = sp.integrate(Iy, (x, a, b))
print(f'I = {Ix}')
```

**Упражнение 2**



```python
x_range = np.linspace(0, 1, 100)
y_range = np.linspace(0, 1, 100)
X, Y = np.meshgrid(x_range, y_range)
f_func = lambda x, y: 1-x-y
Z = f_func(X, Y)
Z = np.where(Z < 0, 0, Z)
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6, antialiased=True)
ax.plot_surface(X, Y, np.zeros_like(Z), color='lightgray', alpha=0.5)
ax.plot_surface(X, np.zeros_like(Y), Z, color='lightgray', alpha=0.5)
ax.plot_surface(np.zeros_like(X), Y, Z, color='lightgray', alpha=0.5)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_zlim(0, 1)
ax.view_init(elev=30, azim=30)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Surface plot of x+y+z = 1')
fig.colorbar(surf, shrink=0.5, aspect=5)
plt.show()

x, y, z = sp.symbols('x y z')
f = z
e, j = (0, 1-x-y)
c, d = (0, 1-x)
a, b = (0, 1)
Iz = sp.integrate(f, (z, e, j))
Iy = sp.integrate(Iz, (y, c, d))
Ix = sp.integrate(Iy, (x, a, b))
print(f'I = {Ix}')
```

**Упражнение 3**





```python
x = np.linspace(1, np.e, 100)
y1 = np.log(x)
y2 = np.zeros_like(x)
x3 = np.ones_like(x)*np.e
y3 = np.linspace(0,1,100)
plt.plot(x, y1, label='y = log(x)')
plt.plot(x, y2, label='y = 0')
plt.plot(x3, y3, label='x = 0')
plt.fill_between(x, y1, y2, where=(y1 > y2), color='lightblue', alpha=0.5)
plt.title('Region between y = log(x) and y = 0')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()

x, y = sp.symbols('x y')
f = (sp.sqrt(4-3*y))/x
c, d = (sp.log(x), 1)
a, b = (1, np.e)
Iy = sp.integrate(f, (y, c, d))
Ix = sp.integrate(Iy, (x, a, b))
print(f'I = {Ix}')
```

**Упражнение 1С**



```python
x = np.linspace(-0.5, 1, 100)
y1 = 1-2*x**2
y2 = -x
plt.plot(x, y1, label='y = 1-2x^2')
plt.plot(x, y2, label='y = -x')
plt.fill_between(x, y1, y2, where=(y1 > y2), color='lightblue', alpha=0.5)
plt.title('Region between y = 1-2x^2 and y = -x')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()

x, y = sp.symbols('x y')
f = sp.sin(x*y)
c, d = (-x, 1-2*x**2)
a, b = (-0.5, 1)
Iy = sp.integrate(f, (y, c, d))
Ix = sp.integrate(Iy, (x, a, b))
print(f'I = {Ix}')
```

**Упражнение 2С**



```python
x = np.linspace(0, 1, 30)
y = np.linspace(0, 1, 30)
X, Y = np.meshgrid(x, y)

Z_roof = X * Y      
Z_floor = np.zeros_like(X) 

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z_floor, color='gray', alpha=0.3, label='z = 0')
ax.plot_surface(X, Y, Z_roof, color='blue', alpha=0.4, label='z = xy')

z_vals = np.linspace(0, 1, 30)
X_c1, Z_c1 = np.meshgrid(x, z_vals)
Y_c1 = X_c1**2
ax.plot_surface(X_c1, Y_c1, Z_c1, color='green', alpha=0.3)

Y_c2, Z_c2 = np.meshgrid(y, z_vals)
X_c2 = Y_c2**2
ax.plot_surface(X_c2, Y_c2, Z_c2, color='red', alpha=0.3)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Объем V: Пол, Крыша и две боковые стенки-цилиндра')
ax.view_init(elev=30, azim=45)
plt.show()

x, y, z = sp.symbols('x y z')
f3 = x * y * z
Iz = sp.integrate(f3, (z, 0, x*y))
Iy = sp.integrate(Iz, (y, x**2, sp.sqrt(x)))
Ix = sp.integrate(Iy, (x, 0, 1))
print(f"Результат Упражнения 2 (второе): {Ix}")
```

**Упражнение 3С**



```python
phi, theta = np.mgrid[0:2*np.pi:20j, 0:np.pi:10j]
x = 0.5 * np.sin(theta) * np.cos(phi)
y = 0.5 * np.sin(theta) * np.sin(phi)
z = 0.5 * np.cos(theta) + 0.5

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, color='cyan', alpha=0.5)
plt.show()

rho, theta, phi = sp.symbols('rho theta phi')
integrand = rho * (rho**2 * sp.sin(theta))
res3 = sp.integrate(integrand, (rho, 0, sp.cos(theta)), 
                   (theta, 0, sp.pi/2), (phi, 0, 2*sp.pi))
print(f"Результат 3: {res3}")
```
